```python
import uproot
import pandas as pd
import matplotlib.pyplot as plt

# 1. ROOT Dosyasını Açalım
dosya_yolu = "/home/yusuf/geant4_work/GEANT4-code/TANK_YENUU/build/electron.root"
dosya = uproot.open(dosya_yolu)

# 2. Dosyanın İçindeki Tree (Ağaç) Yapılarını Görelim
print("--- DOSYA İÇİNDEKİ TREE YAPILARI ---")
print(dosya.keys()) 
# Çıktıda ['TankData;1', 'PhotonData;1'] gibi isimler göreceksiniz.
print("\n" + "="*40 + "\n")

# 3. Tree'lerin İçindeki Branch (Dal/Sütun) İsimlerine Bakalım
tank_tree = dosya["TankData"]
photon_tree = dosya["PhotonData"]

print("--- TankData İÇİNDEKİ BRANCH'LER ---")
print(tank_tree.keys()) # ['Edep', 'Photons']

print("\n--- PhotonData İÇİNDEKİ BRANCH'LER ---")
print(photon_tree.keys()) # ['CherenkovAngle']
print("\n" + "="*40 + "\n")

# 4. Verileri Pandas DataFrame Formatına Çevirelim (Analizi kolaylaştırmak için)
df_tank = tank_tree.arrays(library="pd")
df_photon = photon_tree.arrays(library="pd")

# Terminalde ilk birkaç satırı görerek veriyi kontrol edelim
print("Tank Verileri İlk 5 Satır:")
print(df_tank.head())

print("\nFoton Verileri İlk 5 Satır:")
print(df_photon.head())
print("\n" + "="*40 + "\n")


# =============================================================================
# 5. GRAFİK ÇİZDİRME AŞAMASI
# =============================================================================

# Grafik 1: Enerji Kaybı (Edep) Histogramı
plt.figure(figsize=(8, 5))
plt.hist(df_tank["Edep"], bins=50, color='royalblue', edgecolor='black', alpha=0.7)
plt.title("Detektörde Biriken Enerji Dağılımı (Edep)", fontsize=14)
plt.xlabel("Enerji (MeV)", fontsize=12)
plt.ylabel("Olay Sayısı (Counts)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# Grafik 2: Üretilen Foton Sayısı Histogramı
plt.figure(figsize=(8, 5))
plt.hist(df_tank["Photons"], bins=30, color='darkorange', edgecolor='black', alpha=0.7)
plt.title("Olay Başına Üretilen Optik Foton Sayısı", fontsize=14)
plt.xlabel("Foton Sayısı", fontsize=12)
plt.ylabel("Olay Sayısı", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# Grafik 3: Cherenkov Açısı Dağılımı
plt.figure(figsize=(8, 5))
plt.hist(df_photon["CherenkovAngle"], bins=60, color='crimson', edgecolor='black', alpha=0.7)
plt.title("Cherenkov Işıması Açı Dağılımı", fontsize=14)
plt.xlabel("Cherenkov Açısı (Radyan)", fontsize=12)
plt.ylabel("Foton Sayısı (Counts)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import uproot
import pandas as pd
import matplotlib.pyplot as plt

dosyalar = {
    "100 MeV": "/home/yusuf/geant4_work/GEANT4-code/TANK_YENUU/build/output_100MeV.root",
    "500 MeV": "/home/yusuf/geant4_work/GEANT4-code/TANK_YENUU/build/output_500MeV.root",
    "1 GeV": "/home/yusuf/geant4_work/GEANT4-code/TANK_YENUU/build/output_1GeV.root",
    "1.5 GeV": "/home/yusuf/geant4_work/GEANT4-code/TANK_YENUU/build/output_1.5GeV.root"
}
# Grafik alanını oluşturalım (daha geniş bir alan netlik sağlar)
plt.figure(figsize=(14, 8))

# 2. Her enerji için solid, koyu renkler tanımlayalım
# Stacking order (yığılma sırası): Alttan üste doğru 100, 500, 1000, 1500 MeV
renkler = ['#1f77b4', '#2ca02c', '#ff7f0e', '#d62728'] # Koyu Mavi, Yeşil, Turuncu, Kırmızı
enerji_etiketleri = ["100 MeV", "500 MeV", "1 GeV", "1.5 GeV"]

# 3. Verileri yığılmış (stacked) histogram olarak çizdirelim
# Veri setlerini bir liste halinde hazırlamalıyız
veri_listesi = []
for enerji in enerji_etiketleri:
    dosya_adi = dosyalar[enerji]
    try:
        dosya = uproot.open(dosya_adi)
        # Ntuple adınızın "PhotonData" ve branch adınızın "CherenkovAngle" olduğunu varsayıyorum
        df = dosya["PhotonData"].arrays(library="pd")
        veri_listesi.append(df["CherenkovAngle"])
    except FileNotFoundError:
        print(f"Uyarı: {dosya_adi} bulunamadı.")
        veri_listesi.append(pd.Series(dtype='float64')) # Boş veri seti ekleyerek sırayı koruyalım

# Tek bir hist komutuyla stacked çizim
plt.hist(veri_listesi, 
         bins=100,                  # Daha pürüzsüz görünüm için bin sayısını artırdık
         range=(35, 45),            # Cherenkov pikine odaklanalım
         stacked=True,              # --- İŞTE BURASI: Üst üste yığma ---
         histtype='bar',            # Solid bar grafiği
         color=renkler,             # Liste halindeki renkleri atar
         alpha=1.0,                 # --- İŞTE BURASI: Tam koyu (şeffaf değil) ---
         edgecolor='black',         # Barların etrafına ince siyah çizgi (netlik sağlar)
         linewidth=0.3,             # Çizgi kalınlığı
         label=enerji_etiketleri)  # Lejant etiketleri (stacking sırasına uygun)

# Grafik Süslemeleri (İngilizce yerine Türkçe etiketler)
plt.title("elektron farklı enerjiler için açı ", fontsize=16, fontweight='bold')
plt.xlabel("Cherenkov Açısı (Derece)", fontsize=14)
plt.ylabel("Foton Sayısı", fontsize=14)

# Lejantı (sağ üstteki kutu) güzelleştirelim
plt.legend(fontsize=12, title="Elektron Enerjisi", title_fontsize=13, frameon=True, shadow=True)

# Izgara ekleyelim
plt.grid(True, linestyle='--', alpha=0.5)

# Grafiği göster
plt.tight_layout()
plt.show()
```
